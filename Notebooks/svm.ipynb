# SVM + HOG 

```python
import sklearn as sk 
import numpy as np 
import matplotlib.pyplot as plt 
```

# Load Data

```python
data = np.load("/home/achraf/Research/MLHACK_PAPER/Datasets/cifar10-orig-hog.npz")
X = data["X"]
Y = data["Y"]
print(X.shape)
print(Y.shape)
```

```python
X = X[:1000]
Y = Y[:1000]
```

```python
pipeline = sk.pipeline.Pipeline([
    ('pca', sk.decomposition.PCA(n_components=0.95)),
    ('svm', sk.svm.SVC(max_iter=5000))
])

param_grid = {
    "svm__kernel":['rbf', 'linear', 'poly'],
    "svm__C":[1.0, 0.1, 0.01],
    "svm__gamma": ["auto", "scale"]
}

inner_cv = sk.model_selection.StratifiedKFold(n_splits=5)
grid = sk.model_selection.GridSearchCV(pipeline, param_grid, cv=inner_cv)
outer_cv= sk.model_selection.StratifiedKFold(n_splits=10)
```

```python
nested_scores = sk.model_selection.cross_val_score(grid, X, Y, cv=outer_cv)
print(nested_scores)
```
